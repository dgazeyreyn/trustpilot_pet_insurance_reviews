```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# --- Robust project root resolution ---
# Assumes notebook lives in /notebooks and data lives in /data
PROJECT_ROOT = Path.cwd().resolve().parent
DATA_DIR = PROJECT_ROOT / "data" / "modeling"

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_DIR exists:", DATA_DIR.exists())
```

```python
review_themes_path = DATA_DIR / "reviews_with_themes.csv"

review_themes = pd.read_csv(review_themes_path)

display(review_themes.head())

print("review_themes shape:", review_themes.shape)
```

```python
provider_totals = review_themes.groupby(['provider']).size().sort_values(ascending=False)
print(provider_totals)
```

```python
theme_totals = review_themes.groupby(['theme']).size().sort_values(ascending=False)
print(theme_totals)
```
